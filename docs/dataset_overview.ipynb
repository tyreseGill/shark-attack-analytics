## Dataset Contents

```python
DATA = "../data/attacks.csv"

data_frame = pd.read_csv(
    DATA,
    encoding="latin1",
    usecols=[
        "Date", "Type", "Country", "Area", "Location", "Activity", "Sex ", "Age", "Fatal (Y/N)", "Time", "Species "
    ]
)

data_frame = data_frame.rename(
    columns={
        "Sex ": "Gender",
        "Fatal (Y/N)": "Fatal",
        "Species ": "Species"
    }
)

print(data_frame.fillna("Unknown").head(10))
```
